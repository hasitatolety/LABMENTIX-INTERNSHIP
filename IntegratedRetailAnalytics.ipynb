```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import IsolationForest
from xgboost import XGBRegressor
from sklearn.metrics import mean_squared_error, silhouette_score
import warnings

warnings.filterwarnings('ignore')
sns.set(style='whitegrid')

print('Step 1: Loading datasets...')
try:
    # Loading your specific repo files
    features = pd.read_excel('Features data set.xlsx')
    sales = pd.read_excel('sales data-set.xlsx')
    stores = pd.read_excel('stores data-set.xlsx')
    
    for df_temp in [features, sales, stores]:
        df_temp.columns = df_temp.columns.str.replace(' ', '_').str.title()
        df_temp.rename(columns={'Isholiday': 'IsHoliday', 'Cpi': 'CPI'}, inplace=True)

    df = pd.merge(sales, features, on=['Store', 'Date'], how='left')
    if 'IsHoliday_x' in df.columns:
        df['IsHoliday'] = df['IsHoliday_x']
        df.drop(['IsHoliday_x', 'IsHoliday_y'], axis=1, inplace=True)
    df = pd.merge(df, stores, on='Store', how='left')
    df['Date'] = pd.to_datetime(df['Date'], dayfirst=True, errors='coerce')
    
    print('✅ Data Loaded. Generating Visuals...')
    
    # --- CHART 1: SALES TREND ---
    plt.figure(figsize=(12, 5))
    df.groupby('Date')['Weekly_Sales'].sum().plot(color='blue')
    plt.title('Total Weekly Sales Trend')
    plt.show()

    # --- CHART 2: STORE SIZE VS SALES ---
    plt.figure(figsize=(10, 6))
    sns.scatterplot(data=df.sample(1000), x='Size', y='Weekly_Sales', hue='Type')
    plt.title('Store Size vs Weekly Sales')
    plt.show()
    
    print('✅ Analysis Complete.')
except Exception as e:
    print(f'❌ Error: {e}')
```
